```python
import pandas as pd
import glob
import os
```

```python
files = glob.glob("/content/*.csv")

print("Number of CSV files:", len(files))

for file in files:
    print(os.path.basename(file))
```

```python
df_list = []

for file in files:
    temp_df = pd.read_csv(file)
    df_list.append(temp_df)

df = pd.concat(df_list, ignore_index=True)

print("Dataset shape:", df.shape)
```

We check the 999 value is actually a max value or an invalid value

```python
print("PM2.5 values equal to 999:", (df["PM2.5"] == 999).sum())
print("PM10 values equal to 999:", (df["PM10"] == 999).sum())
print("CO values equal to 999:", (df["CO"] == 999).sum())
print("NO2 values equal to 999:", (df["NO2"] == 999).sum())
print("SO2 values equal to 999:", (df["SO2"] == 999).sum())
print("O3 values equal to 999:", (df["O3"] == 999).sum())
```

```python
df[df["PM2.5"] == 999][["station", "year", "month", "day", "hour", "PM2.5"]].head(20)
```

We will replace the 999 values as we can conclude that they are invalid values and not a maximumvalue of our target variable

```python
pollution_cols = ["PM2.5", "PM10", "CO", "NO2", "SO2", "O3"]

for col in pollution_cols:
    df[col] = df[col].replace(999, pd.NA)
```

```python
for col in pollution_cols:
    print(f"{col} missing values:", df[col].isna().sum())
```
