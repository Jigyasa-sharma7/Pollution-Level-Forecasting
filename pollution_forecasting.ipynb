Pollution Level Forecasting — Dataset Understanding
1. Project Objective

The goal of this project is to predict PM2.5 pollution levels using different environmental, weather, time, and location-related factors.

Target Variable

PM2.5 → The pollution level we want to predict.

Independent Variables / Features

We will use information such as:

Other air pollutants
Weather conditions
Wind conditions
Time of day
Date/season
Monitoring station

The project is therefore a Machine Learning Regression problem because PM2.5 is a continuous numerical value.

2. Dataset Information

The dataset contains hourly air pollution measurements from 12 monitoring stations.

The data covers approximately:

1 March 2013 → 28 February 2017

Our dataset contains approximately:

420,768 observations

Each observation represents pollution and weather conditions recorded at a particular station at a particular hour.

3. Features in the Dataset
A. Pollution Features
PM2.5 — Particulate Matter 2.5

PM2.5 stands for Particulate Matter with a diameter of 2.5 micrometers or smaller.

These are very small particles that can penetrate deep into the respiratory system.

Role in our project:

PM2.5 is our target variable.

In simple words:

We want our machine learning model to predict the PM2.5 value.

PM10 — Particulate Matter 10

PM10 stands for Particulate Matter with a diameter of 10 micrometers or smaller.

PM10 particles are generally larger than PM2.5.

PM10 can be useful for predicting PM2.5 because both represent particulate pollution.

SO2 — Sulfur Dioxide

SO2 stands for Sulfur Dioxide.

It is a gas commonly produced by burning sulfur-containing fuels such as coal and oil.

Possible sources include:

Industries
Power plants
Fuel combustion

SO2 can therefore provide information about pollution from combustion and industrial activities.

NO2 — Nitrogen Dioxide

NO2 stands for Nitrogen Dioxide.

It is commonly produced by combustion processes.

Important sources include:

Vehicle exhaust
Industrial combustion
Power generation

NO2 is particularly useful when studying traffic-related air pollution.

CO — Carbon Monoxide

CO stands for Carbon Monoxide.

It is a colorless and odorless gas mainly produced by incomplete combustion.

Possible sources include:

Vehicle exhaust
Burning fuels
Industrial processes
O3 — Ozone

O3 stands for Ozone.

In this dataset, it refers to ground-level ozone, which is an air pollutant.

Ground-level ozone can form through chemical reactions involving other pollutants, especially in the presence of sunlight.

4. Weather / Meteorological Features
TEMP — Temperature

TEMP represents the air temperature.

The values are approximately measured in:

°C (degrees Celsius)

Temperature can affect:

Atmospheric conditions
Chemical reactions
Pollution formation
Pollution dispersion
PRES — Atmospheric Pressure

PRES represents atmospheric pressure.

It is approximately measured in:

hPa (hectopascals)

Atmospheric pressure can influence weather conditions and the movement or accumulation of pollutants.

DEWP — Dew Point Temperature

DEWP stands for Dew Point Temperature.

The dew point is the temperature at which air becomes saturated with moisture.

In simple terms:

Dew point gives us information about the amount of moisture in the air.

Higher dew point generally means the air contains more moisture.

RAIN — Rainfall

RAIN represents the amount of rainfall/precipitation recorded during the hour.

Rain can remove particles from the atmosphere.

Therefore:

Rainfall can influence PM2.5 concentration.

WSPM — Wind Speed

WSPM represents wind speed.

It is measured in approximately:

meters per second (m/s)

Wind speed is very important for pollution forecasting.

For example:

Low wind speed
      ↓
Pollutants may remain concentrated
      ↓
Higher pollution levels

Whereas:

High wind speed
      ↓
Pollutants can disperse more easily
      ↓
Pollution may decrease
5. Wind Direction
wd — Wind Direction

wd stands for Wind Direction.

It tells us the direction from which the wind is coming.
We currently have 16 directions. N,E,S,W(north,east,south,west)and combination of these

```python
numeric_cols = [
    "CO", "O3", "NO2", "SO2", "PM10",
    "DEWP", "TEMP", "PRES", "RAIN", "WSPM"
]

print(df[numeric_cols].isnull().sum())
```

```python
# Convert columns to numeric
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

What is happening here?

Suppose one station has:

Time-------TEMP
10:00------20
11:00------NaN
12:00------24

Interpolation estimates:

11:00------22

But because we're doing:

groupby("station")

the calculation is separate for each monitoring station.

```python
df[numeric_cols] = (
    df.groupby("station")[numeric_cols]
      .transform(lambda group: group.interpolate(method="linear"))
)
```

```python
print(df[numeric_cols].isnull().sum())
```

Only 22 NO2 values remain missing. These are most likely values at the very beginning or end of a station's time series, where linear interpolation can't work because there isn't a value on both sides.

Let's inspect those 22 rows

```python
df[df["NO2"].isna()][
    ["station", "datetime", "NO2"]
]
```

Linear interpolation couldn't fill them because there is no earlier NO2 value to interpolate from. We should not delete those rows, because 22 out of 420k is tiny and those rows may still contain useful information.
Since they're only at the beginning of a station's data, we can use forward filling within each station.these 22 values are at the very beginning of their station's data, so ffill() has nothing before them to copy from.Better solution: forward fill + backward fill

```python
df["NO2"] = (
    df.groupby("station")["NO2"]
      .transform(lambda x: x.ffill().bfill())
)
```

```python
print("Remaining missing NO2:", df["NO2"].isna().sum())
```

```python
print(df[numeric_cols].isnull().sum())
```

```python
print(df["wd"].value_counts(dropna=False))
```

We don't want to use one direction for the entire dataset because different stations can have different prevailing wind patterns.So if Station A has a missing wd, we'll use Station A's most common wind direction, not the overall most common direction.

```python
station_wind_mode = (
    df.groupby("station")["wd"]
      .agg(lambda x: x.mode()[0] if not x.mode().empty else "N")
)

print(station_wind_mode)
```

```python
df["wd"] = df["wd"].fillna(
    df["station"].map(station_wind_mode)
)
```

```python
print("Missing wind direction:", df["wd"].isna().sum())
print(df["wd"].value_counts())
```

## Data Cleaning Summary

The dataset initially contained 420,768 observations from 12 monitoring stations.

The following preprocessing steps were performed:

1. Combined data from all 12 monitoring stations.
2. Created a `datetime` column using year, month, day, and hour.
3. Sorted observations by station and datetime.
4. Identified invalid `999` values in pollution measurements and replaced them with missing values.
5. Converted pollution and weather variables to appropriate numerical data types.
6. Filled missing numerical measurements using station-wise time-based interpolation.
7. Handled remaining boundary missing values using forward/backward filling within each station.
8. Filled missing wind-direction (`wd`) values using the most frequent wind direction for each station.
9. Removed observations where the target variable `PM2.5` was unavailable.
10. Checked for duplicate observations.

After preprocessing, the final modeling dataset contains:

- 412,028 observations
- 19 columns
- 0 missing values
- 0 duplicate rows

The target variable for the prediction task is `PM2.5`.
