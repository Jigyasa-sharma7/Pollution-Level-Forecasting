What is happening here?

Suppose one station has:

Time-------TEMP
10:00------20

11:00------22

But because we're doing:

groupby("station")

the calculation is separate for each monitoring station.

```python
df[numeric_cols] = (
    df.groupby("station")[numeric_cols]
      .transform(lambda group: group.interpolate(method="linear"))
)
```
